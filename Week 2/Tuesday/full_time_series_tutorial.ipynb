# 📘 Full Guide: Time Series Data Manipulation in Python
Based on Stefan Jansen's methodology for time series analysis using pandas.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

## 1. Working with Dates and Timestamps
Use `pd.Timestamp`, `pd.Period`, and `pd.date_range` to create time-aware data structures.

```python
from datetime import datetime
timestamp = pd.Timestamp(datetime(2023, 1, 1))
print('Timestamp:', timestamp)
print('Year:', timestamp.year)
print('Day Name:', timestamp.day_name())

period = pd.Period('2023-01')
print('Period:', period)
print('Converted to daily:', period.asfreq('D'))
```

## 2. Creating Time Series
You can generate a time-indexed DataFrame or Series using `pd.date_range`.

```python
index = pd.date_range(start='2023-01-01', periods=12, freq='M')
data = np.random.randn(12)
ts = pd.Series(data, index=index)
ts.head()
```

## 3. Time-based Indexing and Slicing
Access sub-periods of a time series using strings.

```python
ts['2023']  # Access all values in 2023
```

## 4. Frequency Conversion with `.asfreq()`
Change the frequency of the data. Useful for upsampling or downsampling without aggregation.

```python
ts_daily = ts.asfreq('D')
ts_daily.head()
```

## 5. Resampling
Use `.resample()` for frequency conversion with aggregation.

```python
ts_resampled = ts.resample('Q').mean()
ts_resampled
```

## 6. Handling Missing Data with `.ffill()` and `.bfill()`

```python
ts_missing = ts_daily.copy()
ts_missing.iloc[5:10] = np.nan
ts_filled = ts_missing.ffill()
ts_bfilled = ts_missing.bfill()
ts_filled.plot(label='Forward Fill')
ts_bfilled.plot(label='Backward Fill')
plt.legend()
plt.title('Forward and Backward Fill')
plt.show()
```

## 7. Reindexing
Align to a new index using `.reindex()` with optional filling.

```python
new_index = pd.date_range('2023-01-01', '2023-04-30', freq='D')
ts_reindexed = ts.reindex(new_index, method='ffill')
ts_reindexed.head()
```

## 8. Normalizing Time Series
Normalize a series to compare growth from a common base.

```python
normalized = ts.div(ts.iloc[0]).mul(100)
normalized.plot(title='Normalized Series')
plt.show()
```

## 9. Shifting and Returns
Use `.shift()` to create lagged series. Calculate returns with `.pct_change()` and `.diff()`.

```python
ts_shift = ts.shift()
returns = ts.pct_change().mul(100)
diffs = ts.diff()
ts.plot(label='Original')
ts_shift.plot(label='Shifted')
plt.legend()
plt.show()
returns.head(), diffs.head()
```

## 10. Rolling and Expanding Statistics

```python
rolling_mean = ts.rolling(window=3).mean()
expanding_mean = ts.expanding().mean()
ts.plot(label='Original')
rolling_mean.plot(label='3-month Rolling Mean')
expanding_mean.plot(label='Expanding Mean')
plt.legend()
plt.title('Rolling and Expanding Averages')
plt.show()
```

## 11. Correlation Between Time Series

```python
# Simulated multivariate time series
df = pd.DataFrame({
    'A': ts,
    'B': ts * 1.5 + np.random.normal(0, 0.1, len(ts)),
    'C': ts * -1 + np.random.normal(0, 0.1, len(ts))
})
correlations = df.pct_change().corr()
sns.heatmap(correlations, annot=True)
plt.title('Correlation Heatmap')
plt.show()
```

## 12. Simulating Time Series Data
Simulate price paths using random returns.

```python
from numpy.random import normal, seed
seed(42)
rand_returns = normal(loc=0, scale=0.01, size=100)
rand_prices = pd.Series(rand_returns).add(1).cumprod().mul(100)
rand_prices.plot(title='Simulated Price Path')
plt.show()
```
